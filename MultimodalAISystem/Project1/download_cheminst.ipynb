```python
import numpy as np
from PIL import Image
import os

# Load dataset
data = np.load("./MultimodalAISystem/chapter1/others/chestmnist.npz")

# Get training images
X_train = data["train_images"]
y_train = data["train_labels"]

# Create output directory
output_dir = "MultimodalAISystem/chapter1/data/chestmnist_images"
os.makedirs(output_dir, exist_ok=True)

# Save images
for i, img in enumerate(X_train):
    image = Image.fromarray(img.astype(np.uint8))
    image.save(os.path.join(output_dir, f"image_{i}.png"))

np.save(os.path.join(output_dir, "train_labels.npy"), y_train)

print(f"Saved {len(X_train)} images to {output_dir}")
```
